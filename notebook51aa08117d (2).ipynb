```python
import os
import pandas as pd

mendeley_path = "/kaggle/input/datasets/dammikawanasinghe/hiru-m-data"

mend_records = []

for root, dirs, files in os.walk(mendeley_path):
    for file in files:
        if file.lower().endswith((".jpg", ".jpeg", ".png", ".webp")):
            full_path = os.path.join(root, file)
            folder = root.split("/")[-1]  # last folder name
            mend_records.append([file, full_path, folder])
```

```python
mend_meta = pd.DataFrame(mend_records, columns=["image_name", "image_path", "folder"])
```

```python
def map_mendeley_label(folder):
    folder = folder.lower()
    if "mcc" in folder:
        return "mcc"
    if "scc" in folder:
        return "scc"
    if "new" in folder:
        return "mel"   # or "bcc" depending on your dataset knowledge
    return "unknown"
```

```python
mend_meta["diagnosis"] = mend_meta["folder"].apply(map_mendeley_label)
```

```python
cancer_labels = ["mel", "melanoma", "bcc", "basal cell carcinoma",
                 "scc", "squamous cell carcinoma", "mcc", "merkel cell carcinoma"]

mend_meta["cancer_binary"] = mend_meta["diagnosis"].apply(
    lambda x: 1 if x in cancer_labels else 0
)
```

```python
shajin_path = "/kaggle/input/datasets/shajinrp/skin-lesion-image"

shajin_records = []

for root, dirs, files in os.walk(shajin_path):
    for file in files:
        if file.lower().endswith((".jpg", ".jpeg", ".png", ".webp")):
            full_path = os.path.join(root, file)
            folder = root.split("/")[-1]
            shajin_records.append([file, full_path, folder])

shajin_meta = pd.DataFrame(shajin_records, columns=["image_name", "image_path", "folder"])
```

```python
def map_shajin_label(folder):
    folder = folder.lower()
    if "mcc" in folder:
        return "mcc"
    if "scc" in folder:
        return "scc"
    if "new" in folder:
        return "mel"   # or "bcc" depending on your dataset knowledge
    return "unknown"
```

```python
shajin_meta["diagnosis"] = shajin_meta["folder"].apply(map_shajin_label)
```

```python
cancer_labels = ["mel", "melanoma", "bcc", "basal cell carcinoma",
                 "scc", "squamous cell carcinoma", "mcc", "merkel cell carcinoma"]

shajin_meta["cancer_binary"] = shajin_meta["diagnosis"].apply(
    lambda x: 1 if x in cancer_labels else 0
)
```
