```python
def map_cancer_label(label):

    if pd.isna(label):
        return None

    label = str(label).strip().lower()

    if "melanoma" in label:
        return 0

    elif "basal cell carcinoma" in label:
        return 1

    elif "squamous cell carcinoma" in label:
        return 2

    elif "merkel cell carcinoma" in label:
        return 3

    else:
        return None
```

```python
def search_mcc(path):

    print("\n====================================")
    print("Searching:", path)
    print("====================================")

    for root, dirs, files in os.walk(path):

        for file in files:

            if file.lower().endswith(
                (".csv", ".txt", ".json")
            ):

                file_path = os.path.join(
                    root,
                    file
                )

                try:

                    if file.lower().endswith(".csv"):

                        temp = pd.read_csv(
                            file_path,
                            low_memory=False
                        )

                        print(
                            "\nFILE:",
                            file_path
                        )

                        for col in temp.columns:

                            if temp[col].dtype == "object":

                                matches = temp[
                                    temp[col]
                                    .astype(str)
                                    .str.contains(
                                        "merkel|mcc",
                                        case=False,
                                        na=False
                                    )
                                ]

                                if len(matches) > 0:

                                    print(
                                        "COLUMN:",
                                        col
                                    )

                                    print(
                                        matches[
                                            [col]
                                        ].head(20)
                                    )

                except Exception as e:

                    print(
                        "Could not read:",
                        file_path
                    )
```

```python
import os
import pandas as pd

SKIN_DATASET = "/kaggle/input/datasets/shajinrp/skin-lesion-image/Dataset"

# ==========================================
# CLASS MAPPING
# ==========================================

class_mapping = {
    "MEL": 0,
    "BCC": 1,
    "SCC": 2,
    "MCC": 3,
    "BKL": 4,
    "DF": 4,
    "NV": 4,
    "VASC": 4
}

class_name_mapping = {
    "MEL": "Melanoma",
    "BCC": "BCC",
    "SCC": "SCC",
    "MCC": "MCC",
    "BKL": "Non-Cancer",
    "DF": "Non-Cancer",
    "NV": "Non-Cancer",
    "VASC": "Non-Cancer"
}

records = []

# ==========================================
# READ ALL FOLDERS
# ==========================================

for folder in sorted(os.listdir(SKIN_DATASET)):

    folder_path = os.path.join(
        SKIN_DATASET,
        folder
    )

    if not os.path.isdir(folder_path):
        continue

    # Ignore AK for now
    if folder == "AK":
        print("Skipping AK")
        continue

    if folder not in class_mapping:
        continue

    for file in os.listdir(folder_path):

        if file.lower().endswith(
            (".jpg", ".jpeg", ".png")
        ):

            records.append({
                "image_path": os.path.join(
                    folder_path,
                    file
                ),
                "image_name": file,
                "original_class": folder,
                "class_name": class_name_mapping[folder],
                "label": class_mapping[folder],
                "dataset": "Skin-Lesion-Image"
            })

# ==========================================
# CREATE DATAFRAME
# ==========================================

dataset_df = pd.DataFrame(records)

print("Total images:", len(dataset_df))

print("\nClass distribution:")
print(dataset_df["class_name"].value_counts())

print("\nOriginal folder distribution:")
print(dataset_df["original_class"].value_counts())
```

```python
# ============================================================
# STEP 27 - CONTINUE AFTER y_true ERROR
# ============================================================

import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)
from sklearn.preprocessing import label_binarize

print("=" * 60)
print("STEP 27 - CONTINUING FINAL CLASSIFICATION VERIFICATION")
print("=" * 60)

# ------------------------------------------------------------
# 1. FIX TRUE LABEL FORMAT
# ------------------------------------------------------------

# test_generator.classes may be a Python list
y_true = np.array(test_generator.classes)

# y_prob should already exist from previous part
# If it does not exist, generate it again.
if "y_prob" not in globals():

    print("\nGenerating test predictions...")

    test_generator.reset()

    y_prob = final_model.predict(
        test_generator,
        verbose=1
    )

# Convert predictions to NumPy array
y_prob = np.asarray(y_prob)

# Predicted class
y_pred = np.argmax(y_prob, axis=1)

print("\nPrediction generation completed!")

print("\nTrue label shape:")
print(y_true.shape)

print("\nPrediction probability shape:")
print(y_prob.shape)

print("\nPredicted label shape:")
print(y_pred.shape)

# ------------------------------------------------------------
# 2. CHECK SAMPLE COUNT
# ------------------------------------------------------------

print("\nNumber of test images:", len(y_true))

if len(y_true) != len(y_pred):
    raise ValueError(
        f"ERROR: Number mismatch! "
        f"True labels = {len(y_true)}, "
        f"Predictions = {len(y_pred)}"
    )

print("Sample count verified successfully!")

# ------------------------------------------------------------
# 3. CLASS NAMES
# ------------------------------------------------------------

class_names = [
    "BCC",
    "MCC",
    "Melanoma",
    "Non-Cancer",
    "SCC"
]

# ------------------------------------------------------------
# 4. FINAL ACCURACY
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_true,
    y_pred
)

print("\n" + "=" * 60)
print("FINAL TEST ACCURACY")
print("=" * 60)

print(f"Accuracy : {accuracy:.4f}")
print(f"Accuracy : {accuracy * 100:.2f}%")

# ------------------------------------------------------------
# 5. CLASSIFICATION REPORT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL CLASSIFICATION REPORT")
print("=" * 60)

report = classification_report(
    y_true,
    y_pred,
    labels=np.arange(len(class_names)),
    target_names=class_names,
    digits=4,
    zero_division=0
)

print(report)

# ------------------------------------------------------------
# 6. CONFUSION MATRIX
# ------------------------------------------------------------

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=np.arange(len(class_names))
)

print("\n" + "=" * 60)
print("FINAL CONFUSION MATRIX")
print("=" * 60)

print(cm)

# ------------------------------------------------------------
# 7. CONFUSION MATRIX VISUAL
# ------------------------------------------------------------

plt.figure(figsize=(8, 7))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=class_names,
    yticklabels=class_names,
    cmap="Blues"
)

plt.title("DERM-AI Final Test Set Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("True Class")
plt.tight_layout()

cm_path = "/kaggle/working/STEP27_final_confusion_matrix.png"

plt.savefig(
    cm_path,
    dpi=120,
    bbox_inches="tight"
)

plt.show()

print("\nConfusion matrix saved:")
print(cm_path)

# ------------------------------------------------------------
# 8. ROC / AUC
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL ROC-AUC ANALYSIS")
print("=" * 60)

y_true_bin = label_binarize(
    y_true,
    classes=np.arange(len(class_names))
)

auc_scores = {}

for i, class_name in enumerate(class_names):

    try:

        auc_value = roc_auc_score(
            y_true_bin[:, i],
            y_prob[:, i]
        )

        auc_scores[class_name] = auc_value

        print(
            f"{class_name:<15}: "
            f"AUC = {auc_value:.4f}"
        )

    except ValueError:

        auc_scores[class_name] = np.nan

        print(
            f"{class_name:<15}: "
            f"AUC could not be calculated"
        )

# Macro AUC
macro_auc = roc_auc_score(
    y_true_bin,
    y_prob,
    average="macro",
    multi_class="ovr"
)

print("\n" + "-" * 60)

print(f"Macro AUC : {macro_auc:.4f}")
print(f"Macro AUC : {macro_auc * 100:.2f}%")

# ------------------------------------------------------------
# 9. TRUE CLASS DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("TRUE TEST CLASS DISTRIBUTION")
print("=" * 60)

for i, class_name in enumerate(class_names):

    count = np.sum(y_true == i)

    print(
        f"{class_name:<15}: {count}"
    )

# ------------------------------------------------------------
# 10. PREDICTED CLASS DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("PREDICTED CLASS DISTRIBUTION")
print("=" * 60)

for i, class_name in enumerate(class_names):

    count = np.sum(y_pred == i)

    print(
        f"{class_name:<15}: {count}"
    )

# ------------------------------------------------------------
# 11. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL DERM-AI CLASSIFICATION PERFORMANCE")
print("=" * 60)

print(f"Model              : EfficientNetB0")
print(f"Test Samples       : {len(y_true)}")
print(f"Test Accuracy      : {accuracy * 100:.2f}%")
print(f"Macro AUC          : {macro_auc * 100:.2f}%")

print("\nPer-Class AUC:")

for class_name, value in auc_scores.items():

    if not np.isnan(value):
        print(
            f"{class_name:<15}: {value:.4f}"
        )

print("\n" + "=" * 60)
print("STEP 27 COMPLETED SUCCESSFULLY!")
print("=" * 60)
```

```python
# ============================================================
# STEP 28 - CANCER vs NON-CANCER BINARY EVALUATION
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    roc_auc_score
)

print("=" * 60)
print("STEP 28 - CANCER vs NON-CANCER BINARY EVALUATION")
print("=" * 60)

# ------------------------------------------------------------
# CLASS MAPPING
# ------------------------------------------------------------

class_names = [
    "BCC",
    "MCC",
    "Melanoma",
    "Non-Cancer",
    "SCC"
]

# Non-Cancer = class 3
# Everything else = Cancer

print("\nOriginal classes:")
for i, name in enumerate(class_names):
    print(f"{i}: {name}")

# ------------------------------------------------------------
# TRUE LABELS -> BINARY
# ------------------------------------------------------------

# 0 = Non-Cancer
# 1 = Cancer

y_true_binary = np.where(
    y_true == 3,
    0,
    1
)

# ------------------------------------------------------------
# PREDICTED PROBABILITY -> BINARY
# ------------------------------------------------------------

# Cancer probability =
# BCC + MCC + Melanoma + SCC

cancer_probability = (
    y_prob[:, 0] +
    y_prob[:, 1] +
    y_prob[:, 2] +
    y_prob[:, 4]
)

# Non-Cancer probability
non_cancer_probability = y_prob[:, 3]

# Binary prediction using maximum probability
y_pred_binary = np.where(
    cancer_probability >= non_cancer_probability,
    1,
    0
)

# ------------------------------------------------------------
# DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("BINARY CLASS DISTRIBUTION")
print("=" * 60)

print("\nTRUE:")
print("Non-Cancer :", np.sum(y_true_binary == 0))
print("Cancer     :", np.sum(y_true_binary == 1))

print("\nPREDICTED:")
print("Non-Cancer :", np.sum(y_pred_binary == 0))
print("Cancer     :", np.sum(y_pred_binary == 1))

# ------------------------------------------------------------
# ACCURACY
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_true_binary,
    y_pred_binary
)

# ------------------------------------------------------------
# PRECISION
# ------------------------------------------------------------

precision = precision_score(
    y_true_binary,
    y_pred_binary,
    zero_division=0
)

# ------------------------------------------------------------
# RECALL
# ------------------------------------------------------------

recall = recall_score(
    y_true_binary,
    y_pred_binary,
    zero_division=0
)

# ------------------------------------------------------------
# F1
# ------------------------------------------------------------

f1 = f1_score(
    y_true_binary,
    y_pred_binary,
    zero_division=0
)

print("\n" + "=" * 60)
print("CANCER vs NON-CANCER PERFORMANCE")
print("=" * 60)

print(f"Accuracy  : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Precision : {precision:.4f} ({precision*100:.2f}%)")
print(f"Recall    : {recall:.4f} ({recall*100:.2f}%)")
print(f"F1-Score  : {f1:.4f} ({f1*100:.2f}%)")

# ------------------------------------------------------------
# CLASSIFICATION REPORT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("BINARY CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_true_binary,
        y_pred_binary,
        target_names=[
            "Non-Cancer",
            "Cancer"
        ],
        digits=4,
        zero_division=0
    )
)

# ------------------------------------------------------------
# CONFUSION MATRIX
# ------------------------------------------------------------

binary_cm = confusion_matrix(
    y_true_binary,
    y_pred_binary
)

print("\n" + "=" * 60)
print("BINARY CONFUSION MATRIX")
print("=" * 60)

print(binary_cm)

# ------------------------------------------------------------
# CONFUSION MATRIX VISUAL
# ------------------------------------------------------------

plt.figure(figsize=(7, 6))

plt.imshow(
    binary_cm,
    interpolation="nearest"
)

plt.title(
    "DERM-AI Cancer vs Non-Cancer Confusion Matrix"
)

plt.colorbar()

plt.xticks(
    [0, 1],
    ["Non-Cancer", "Cancer"]
)

plt.yticks(
    [0, 1],
    ["Non-Cancer", "Cancer"]
)

plt.xlabel("Predicted Class")
plt.ylabel("True Class")

for i in range(2):
    for j in range(2):

        plt.text(
            j,
            i,
            str(binary_cm[i, j]),
            ha="center",
            va="center",
            fontsize=16
        )

plt.tight_layout()

cm_path = (
    "/kaggle/working/"
    "STEP28_cancer_vs_non_cancer_confusion_matrix.png"
)

plt.savefig(
    cm_path,
    dpi=120,
    bbox_inches="tight"
)

plt.show()

print("\nConfusion matrix saved:")
print(cm_path)

# ------------------------------------------------------------
# ROC + AUC
# ------------------------------------------------------------

fpr, tpr, thresholds = roc_curve(
    y_true_binary,
    cancer_probability
)

binary_auc = roc_auc_score(
    y_true_binary,
    cancer_probability
)

print("\n" + "=" * 60)
print("BINARY ROC-AUC")
print("=" * 60)

print(f"Binary AUC : {binary_auc:.4f}")
print(f"Binary AUC : {binary_auc*100:.2f}%")

# ------------------------------------------------------------
# ROC CURVE VISUAL
# ------------------------------------------------------------

plt.figure(figsize=(8, 7))

plt.plot(
    fpr,
    tpr,
    linewidth=2,
    label=f"Cancer vs Non-Cancer (AUC = {binary_auc:.4f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title(
    "DERM-AI Cancer vs Non-Cancer ROC Curve"
)

plt.legend(
    loc="lower right"
)

plt.grid(True, alpha=0.3)

plt.tight_layout()

roc_path = (
    "/kaggle/working/"
    "STEP28_cancer_vs_non_cancer_ROC.png"
)

plt.savefig(
    roc_path,
    dpi=120,
    bbox_inches="tight"
)

plt.show()

print("\nROC curve saved:")
print(roc_path)

# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("STEP 28 - FINAL BINARY CLASSIFICATION SUMMARY")
print("=" * 60)

print("Model            : EfficientNetB0")
print("Task             : Cancer vs Non-Cancer")
print(f"Test Samples     : {len(y_true_binary)}")
print(f"Accuracy         : {accuracy*100:.2f}%")
print(f"Precision        : {precision*100:.2f}%")
print(f"Recall           : {recall*100:.2f}%")
print(f"F1-Score         : {f1*100:.2f}%")
print(f"Binary AUC       : {binary_auc*100:.2f}%")

print("\nVisuals generated:")
print("1. Cancer vs Non-Cancer Confusion Matrix")
print("2. Cancer vs Non-Cancer ROC Curve")

print("\n" + "=" * 60)
print("STEP 28 COMPLETED SUCCESSFULLY!")
print("=" * 60)
```

```python
import os
import shutil
import json

# ============================================================
# DERMAI CLASSIFICATION - FINAL BACKUP
# ============================================================

BACKUP_DIR = "/kaggle/working/DERMAI_Classification_Results"

# Create backup folder
os.makedirs(BACKUP_DIR, exist_ok=True)

# Files we want to preserve
files_to_save = [
    "/kaggle/working/DERMAI_final_classification_model.keras",
    "/kaggle/working/best_dermai_model_v2.keras",
    "/kaggle/working/DERMAI_class_names.json",
    "/kaggle/working/DERMAI_final_classification_results.json",
    "/kaggle/working/dermai_training_metadata.csv",
    "/kaggle/working/STEP27_final_confusion_matrix.png",
    "/kaggle/working/STEP28_cancer_vs_non_cancer_confusion_matrix.png",
    "/kaggle/working/STEP28_cancer_vs_non_cancer_ROC.png",
]

# Copy existing files
copied = []

for file_path in files_to_save:
    if os.path.exists(file_path):
        shutil.copy2(file_path, BACKUP_DIR)
        copied.append(os.path.basename(file_path))
    else:
        print("Not found:", file_path)

print("\n========== BACKUP COMPLETE ==========")
print("Files copied:", len(copied))

for f in copied:
    print("✓", f)

# Create a simple README
readme_path = os.path.join(BACKUP_DIR, "README.txt")

with open(readme_path, "w") as f:
    f.write("DERMAI Classification Model Backup\n")
    f.write("=================================\n\n")
    f.write("Architecture: EfficientNetB0 Transfer Learning\n")
    f.write("Classification Classes:\n")
    f.write("1. BCC\n")
    f.write("2. MCC\n")
    f.write("3. Melanoma\n")
    f.write("4. Non-Cancer\n")
    f.write("5. SCC\n\n")
    f.write("Main Model:\n")
    f.write("DERMAI_final_classification_model.keras\n\n")
    f.write("Stage Prediction:\n")
    f.write("Not included. Stage prediction will be developed separately.\n")

print("\nREADME created.")
```
